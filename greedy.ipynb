```python
def nearestNeighboor(startPoint, distances, Pathlength = 0, maxDistance = 9999):
    
    if (Pathlength != 0 & Pathlength != len(distances[0][:])):
        print("Wrong distance matrix given for this number of points")

    else:

        assumedMaxDist = 10000
        n = len(distances[0][:])
        currentCity = startPoint-1
        totalWeight = 0
        shortestPath_greedy = np.zeros(n)

        for k in range(n):
            distances[k][k] = assumedMaxDist

        if (maxDistance > 10000):
            print('The maximal Distance is too large. Please modify code')

        for i in range(n):
            shortestPath_greedy[i] = currentCity+1
            nearestCity = np.argmin(distances[currentCity][:])
            totalWeight += distances[currentCity][nearestCity]
            distances[currentCity][nearestCity] = assumedMaxDist
            distances[nearestCity][currentCity] = assumedMaxDist
            currentCity = nearestCity
            #print(totalWeight)
        
        shortestPath_greedy = shortestPath_greedy.astype(int)

    return(totalWeight, shortestPath_greedy)

def nearestNeighboor_short(startPoint, distances):

    assumedMaxDist = 10000
    n = len(distances[0][:])
    currentCity = startPoint-1
    totalWeight = 0
    shortestPath_greedy = np.zeros(n)

    for k in range(n):
        distances[k][k] = assumedMaxDist

    for i in range(n):
        shortestPath_greedy[i] = currentCity+1
        nearestCity = np.argmin(distances[currentCity][:])
        totalWeight += distances[currentCity][nearestCity]
        distances[currentCity][nearestCity] = assumedMaxDist
        distances[nearestCity][currentCity] = assumedMaxDist
        currentCity = nearestCity
        #print(totalWeight)
    
    shortestPath_greedy = shortestPath_greedy.astype(int)

    return(totalWeight, shortestPath_greedy)


def nearestNeighboor_allStarts(distances, Pathlength = 0, maxDistance = 9999):
    
    if (Pathlength != 0 & Pathlength != len(distances[0][:])):
        print("Wrong distance matrix given for this number of points")

    if (maxDistance > 10000):
            print('The maximal Distance is too large. Please modify code')

    else:
        assumedMaxDist = 10000
        n = len(distances[0][:])
        totalWeights = np.zeros(n)
        shortestPaths_greedy = np.zeros((n,n))

        for l in range(n):
            totalWeights[l], shortestPaths_greedy[l][:] = nearestNeighboor_short(l, distances)

    return(totalWeights, shortestPaths_greedy)
```
